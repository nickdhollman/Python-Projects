## 1. Create .venv_iris

Run this cell to create .venv_iris inside the project folder.

The Windows Python launcher command py -3.12 selects Python 3.12. The -m option runs its venv module. Install the project packages in the next step.


```python
from pathlib import Path
import subprocess

project_dir = Path.cwd().resolve()
if not (project_dir / "requirements.txt").is_file():
    raise RuntimeError("Open the 01_iris_classification project folder before running setup.")

venv_dir = project_dir / ".venv_iris"
venv_python = venv_dir / "Scripts" / "python.exe"

if venv_python.is_file():
    print("The project environment already exists.")
else:
    subprocess.run(
        ["py", "-3.12", "-m", "venv", str(venv_dir)],
        check=True,
    )
    print("Created the project environment", venv_dir)
```

## 2. Install the requirements

Run this cell after the environment creation cell finishes to install the packages listed in requirements.txt.

The command runs pip through the Python executable inside .venv_iris, which makes the installation target explicit. The -r option reads the requirements file. Pip downloads the listed packages and their supporting dependencies. This step requires internet access.


```python
subprocess.run(
    [str(venv_python), "-m", "pip", "install", "-r", str(project_dir / "requirements.txt")],
    check=True,
)
```

## 3. Verify the environment

Run this cell to display the environment's Python version and check its declared package dependencies.

Expect Python 3.12.x and No broken requirements found.


```python
subprocess.run([str(venv_python), "--version"], check=True)
subprocess.run([str(venv_python), "-m", "pip", "check"], check=True)
```

## 4. Open the Iris notebook

After installation and verification succeed, open 01_Iris_Project.ipynb. Select this project's .venv_iris as its kernel. Browse to .venv_iris/Scripts/python.exe if needed.

Run the Iris notebook from the beginning.

The local .env contains RANDOM_STATE and TEST_SIZE. When starting from a fresh copy of the project, copy .env.example to .env before running the analysis.
