# Create a Project Environment

Use this template to create a separate Python environment for each project.

1. Create a project folder and add requirements.txt and a safe .env.example.
2. Open this notebook with an existing working Python kernel.
3. Change the project settings below.
4. Run the setup cells in order.
5. Open the project's analysis notebook and select the new environment as its kernel.

This template uses only Python's standard library. Package installation happens inside the new environment. Installation requires internet access when packages are not already available locally.

If no working notebook kernel is available, use the terminal setup instructions in README.md first. A notebook needs a kernel before it can execute setup code.

## 1. Choose the project settings

Change PROJECT_FOLDER, ENVIRONMENT_NAME and PYTHON_VERSION for the project.

A relative project path is resolved from the working directory printed by this cell. For example, use Path("01_iris_classification") when the working directory is projects. Use an absolute Path if working from a different folder.

Use an environment name starting with .venv, such as .venv_iris or .venv_customer_churn. The repository's .gitignore already excludes folders with that prefix.

The selected Python version must already be installed. This template creates environments from an existing Python installation.

```python
from pathlib import Path
import os
import subprocess

# Change these three settings for each project.
PROJECT_FOLDER = Path("your_project_folder")
ENVIRONMENT_NAME = ".venv_your_project"
PYTHON_VERSION = "3.12"

print("Current working directory", Path.cwd())

project_dir = PROJECT_FOLDER.expanduser().resolve()
requirements_file = project_dir / "requirements.txt"

if not project_dir.is_dir():
    raise FileNotFoundError("Create the project folder or correct PROJECT_FOLDER.")
if not requirements_file.is_file():
    raise FileNotFoundError("Add requirements.txt to the project folder before setup.")
if not ENVIRONMENT_NAME.startswith(".venv") or Path(ENVIRONMENT_NAME).name != ENVIRONMENT_NAME:
    raise ValueError("Use a single folder name starting with .venv.")

venv_dir = project_dir / ENVIRONMENT_NAME
venv_python = venv_dir / ("Scripts/python.exe" if os.name == "nt" else "bin/python")
python_command = ["py", f"-{PYTHON_VERSION}"] if os.name == "nt" else [f"python{PYTHON_VERSION}"]

print("Target project", project_dir)
print("Target environment", venv_dir)
```

## 2. Create the environment

Run this cell to create the selected environment.

On Windows, py selects the installed Python version. On macOS or Linux, the command uses python followed by the version, such as python3.12. The -m venv arguments run Python's environment creation module.

An existing environment is reused after checking its interpreter version. The cell reports an incomplete environment directory instead of overwriting it. To change Python versions, create a separately named environment and install the dependencies there.

```python
if venv_dir.exists():
    if not venv_python.is_file():
        raise FileExistsError("The target directory exists but has no environment interpreter. Review it before retrying.")
    print("Using the existing project environment.")
else:
    subprocess.run([*python_command, "-m", "venv", str(venv_dir)], check=True)
    print("Created the project environment.")

version_check = subprocess.run(
    [str(venv_python), "-c", "import sys; print(f'{sys.version_info.major}.{sys.version_info.minor}')"],
    check=True,
    capture_output=True,
    text=True,
)
if version_check.stdout.strip() != PYTHON_VERSION:
    raise RuntimeError("The environment's Python version differs from PYTHON_VERSION. Choose a new environment name.")
print("Environment Python version", version_check.stdout.strip())
```

## 3. Install the project requirements

Run this cell after environment creation succeeds.

The command invokes pip through the new environment's Python executable. The -r option reads requirements.txt. Supporting dependencies are installed automatically.

Review requirements.txt before running the cell. Keep authentication tokens out of dependency URLs and the notebook. If a private package source is required, follow the approved credential setup for that project.

```python
subprocess.run(
    [str(venv_python), "-m", "pip", "install", "-r", str(requirements_file)],
    check=True,
)
```

## 4. Verify the installation

Run this cell to confirm the interpreter version and check declared package dependencies.

A successful pip check reports No broken requirements found. After selecting the new kernel, run the project's imports and analysis to verify its actual behavior as well.

```python
subprocess.run([str(venv_python), "--version"], check=True)
subprocess.run([str(venv_python), "-m", "pip", "check"], check=True)
```

## 5. Create local configuration from the safe example

Run this cell to copy .env.example to .env if the local file is absent. Existing .env contents are preserved.

Edit the local .env with the project's configuration afterward. Keep private values out of .env.example. The cell reports filenames and status without displaying configuration values.

```python
import shutil

example_file = project_dir / ".env.example"
local_env = project_dir / ".env"

if local_env.exists():
    print("Local .env already exists.")
elif example_file.is_file():
    shutil.copyfile(example_file, local_env)
    print("Created local .env from .env.example.")
else:
    print("Add a safe .env.example if this project needs local configuration.")
```

## 6. Select the environment for the project notebook

Open the project's analysis notebook and choose Select Kernel, then select the environment created above. Browse to its Python executable if needed.

Run the following verification in the analysis notebook.

~~~python
import sys
print(sys.executable)
~~~

Confirm that the path includes the project's chosen environment name before importing the analysis packages. Run the project from its own folder so relative configuration and data paths resolve as intended.

Create an environment once for each project. Reuse it on later visits. Recreate it from the dependency files when needed instead of copying the environment folder between machines.
